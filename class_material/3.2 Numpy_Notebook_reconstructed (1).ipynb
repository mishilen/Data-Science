## 1. Import NumPy

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. NumPy arrays (`ndarray`)

NumPy's main data structure is the multidimensional `ndarray`, designed for efficient numerical operations.


```python
a = np.array([10, 20, 30., 40, 50])

print(a)
print("ndim:", a.ndim)
print("shape:", a.shape)
print("size:", a.size)
print("dtype:", a.dtype)
```

```python
b = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(b)
print("ndim:", b.ndim)
print("shape:", b.shape)
print("size:", b.size)
print("dtype:", b.dtype)
```

## 3. Creating arrays

```python
zeros = np.zeros((3, 4))
ones = np.ones((2, 3))
sequence = np.arange(0, 10, 2)
evenly_spaced = np.linspace(0, 1, 5)

print("zeros:\n", zeros)
print("\nones:\n", ones)
print("\narange:", sequence)
print("\nlinspace:", evenly_spaced)
```

## 4. Indexing and slicing

```python
x = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("x:\n", x)
print("First row:", x[0])
print("Second column:", x[:, 1])
print("Element at row 2, column 3:", x[1, 2])
print("Top-left 2x2 block:\n", x[:2, :2])
print("Last row:", x[-1])
```

## 5. Boolean and fancy indexing

```python
values = np.array([3, 8, 2, 10, 5, 7])

indices = values > 5
print(indices)
print("Values greater than 5:", values[indices])

indices = np.array([0, 3, 5])
print("Selected positions:", values[indices])
```

## 6. Data types

NumPy arrays have a `dtype`. Explicit data types are useful when memory usage or numerical representation matters.


```python
integers = np.array([1, 2, 3], dtype=np.int32)
floats = np.array([1, 2, 3], dtype=np.float64)

print(integers, integers.dtype)
print(floats, floats.dtype)

converted = integers.astype(np.float64)
print("Converted:", converted, converted.dtype)
```

## 7. Arithmetic operations

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

print("u + v =", u + v)
print("u - v =", u - v)
print("u * v =", u * v)
print("u / v =", u / v)
print("u squared =", u ** 2)
print("2 * u =", 2 * u)
```

## 8. Aggregation and mathematical operations

```python
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("sum:", data.sum())
print("mean:", data.mean())
print("minimum:", data.min())
print("maximum:", data.max())
print("column sums:", data.sum(axis=0))
print("row sums:", data.sum(axis=1))
```

## 9. Broadcasting

Broadcasting allows NumPy to perform arithmetic between arrays with compatible shapes without explicitly replicating the smaller array.


```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

offset = np.array([10, 20, 30])
result = matrix + offset

print("matrix:\n", matrix)
print("offset:", offset)
print("result:\n", result)
```

## 10. Reshaping

```python
numbers = np.arange(1, 13)
matrix = numbers.reshape(3, 4)

print("Original:", numbers)
print("Reshaped:\n", matrix)
print("Shape:", matrix.shape)
```

## 11. NumPy versus Python lists

Vectorized NumPy operations can avoid an explicit Python loop for many numerical calculations. This cell gives a small reproducible comparison.


```python
import time

n = 1_000_000
python_list = list(range(n))
numpy_array = np.arange(n)

start = time.perf_counter()
python_result = [x * 2 for x in python_list]
python_time = time.perf_counter() - start

start = time.perf_counter()
numpy_result = numpy_array * 2
numpy_time = time.perf_counter() - start

print(f"Python list comprehension: {python_time:.6f} seconds")
print(f"NumPy vectorized operation: {numpy_time:.6f} seconds")
print("Results equal:",
      np.array_equal(np.asarray(python_result), numpy_result))
```

## 12. WIFIRE-style satellite image analysis

The archived course material is publicly documented as using a WIFIRE satellite-image example to demonstrate NumPy array processing. A satellite image can be represented as a multidimensional array in which rows and columns correspond to pixels and channels contain image information.

The following synthetic example demonstrates the same NumPy operation without requiring the original external WIFIRE dataset.


```python
# Synthetic RGB image: height x width x channels
image = np.array([
    [[10, 20, 30], [40, 50, 60], [70, 80, 90]],
    [[15, 25, 35], [45, 55, 65], [75, 85, 95]],
    [[20, 30, 40], [50, 60, 70], [80, 90, 100]]
], dtype=np.uint8)

print("Image shape:", image.shape)
print("Rows:", image.shape[0])
print("Columns:", image.shape[1])
print("Channels:", image.shape[2])
```

```python
red = image[:, :, 0]
green = image[:, :, 1]
blue = image[:, :, 2]

print("Red channel:\n", red)
print("\nGreen channel:\n", green)
print("\nBlue channel:\n", blue)
```

```python
brightness = image.mean(axis=2)
print("Average brightness per pixel:\n", brightness)
```
