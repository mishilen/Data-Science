# 2. Data Frame
<p>A DataFrame represents a rectangular table of data and contains an ordered, named collection of <br>
columns, each of which can be a different value type (numeric, string, Boolean, etc.). <br>
The DataFrame has both a row and column index; <br>
it can be thought of as a dictionary of Series all sharing the same index</p>

```python
import pandas as pd
```

```python
data = {"state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],
"year": [2000, 2001, 2002, 2001, 2002, 2003],
"pop": [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}

frame = pd.DataFrame(data)
```

<p>The resulting DataFrame will have its index assigned automatically, as with Series, <br>
and the columns are placed according to the order of the keys in data (which depends on their <br>
insertion order in the dictionary)<p>

```python
frame
```

the <b>head</b> method selects only the first five rows
the <b>tail</b> returns the last five rows

```python
frame.head(), frame.tail()
```

# Altering Columns
If you specify a sequence of columns, the DataFrame’s columns will be arranged in that order

```python
pd.DataFrame(data, columns=["pop", "state", "year"])
```

## Passing a column not in dictionary
<p>Results in missing value (NaN)</p>

```python
frame2 = pd.DataFrame(data, columns=["year", "state", "pop", "debt"])

frame2
```

```python
frame2.columns
```

# Accessing Columns
<p>A column in a DataFrame can be retrieved as a Series either by <b>dictionary-like </b>notation
<br>
or by using the <b>dot attribute</b> notation</p>

```python
frame2.state, frame2["state"]
```

## Loc and Iloc

```python
frame2.loc[1], frame2.iloc[1]
```

```python
frame.columns = ["Province", "Occur", "Num"]

frame
```

```python
import pandas as pd
import numpy as np

s = pd.Series(['a', 3, np.nan, 1, np.nan])

print(s.notnull().sum())
```

# Assigning values
## List or arrays
<p>When you are assigning lists or arrays to a column, the value’s length must match <br>
the length of the DataFrame. </p>
<p>If you assign a Series, its labels will be realigned exactly to the DataFrame’s index, <br>
insertingmissing values in any index values not present</p>

```python
frame2["debt"] = [1.1, 3.1, 200, 500, 2, 1] # list or array is exact
frame2
```

## Creating new column
<p>Assigning a column that doesn’t exist will create a new column.</p>

```python
pay = pd.Series([200, 100], index=[5, 0])
frame2["Pay"] = pay
frame, frame2
```

## Deleting Column

```python
del frame2["Pay"]
```
