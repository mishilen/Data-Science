# Importing Pandas and Numpy

```python
import numpy as np
import pandas as pd
```

# Data Sctructures

# 1. Series
<p>A Series is a one-dimensional array-like object containing a sequence of values (of similar types to
<br>
NumPy types) of the same type and an associated array of data labels, called its index. The simplest
<br>
Series is formed from only an array of data</p>

```python
data = [1, 5, 7, 11]
series0 = pd.Series(data)
series0
```

<h2>Converting numpy array to pandas series</h2>

```python
arr = np.array(data)
pd.Series(arr)
```


<p>Get the array representation and index object of the Series via its 
<br>
<code>array</code> and <code>index</code> attributes</p>

```python
series0.array, series0.index
```

## Creating a Series with an index identifying each data point with a label

```python
series1 = pd.Series(data, index=['a', 'b', 'c', 'd'])
series1
```

<p>Compared with NumPy arrays, you can use labels in the index when selecting single values or 
<br>a set of
values</p>

```python
series1['a'], series1['d'], series1[['a', 'd']], "str index", series1['a': 'c'], "numeric index", series0[0:4]
```

## Boolean Indexing and Scalar Arithmetic

```python
series0[series0 > 5], series0 * 2
```

```python
"b" in series1, "e" in series1, 5 in series1, 5 in series1.array
```

## Dictionary to Series

```python
sdata = {"Ohio": 35000, "Texas": 71000, "Oregon": 16000, "Utah": 5000}

series2 = pd.Series(sdata)

series2, series2.index, series2.array, series2["Ohio"]
```

## Overriding default keys
<p>passing an index with the dictionary keys in the order you want them to appear in the
<br>resulting Series</p>

```python
indexes = ["LA", "Ohio", "Oregon", "Texas"]

series3 = pd.Series(sdata, index=indexes)

series3
```

<p>Since no Value was found for <b>LA</b> it was replaced with <b>NaN</b></p>

## isna & notna functions and methods

```python
pd.isna(series3), pd.notna(series3)
```

```python
series3.isna(), series3.notna()
```

## Series Arithmetic

```python
series2 + series3 # incompatible indexes were replaced by NaN
```

<p>Both the Series object itself and its index have a <b>name</b> attribute</p>

```python
series3.name = "Population"
series3.index.name = "States"

series3
```

## Altering indexes

```python
series3.index = ["Hyderabad", "Karachi", "Islamabad", "Lahore"]
series3
```

```python
series3['Karachi'], series3.loc['Karachi'], series3.iloc[1]
```

```python
inputs = {"Day 1" : [1, 4, 5],
            "Day 2" : [6, 1, 2]}

series4 = pd.Series(inputs)
df = pd.DataFrame(inputs)

series4, df
```

```python
series5 = pd.Series(["Bulbasaur", "Ivysaur", "Charmander"], index=range(1, 6, 2))

series5
```
