## 8. Create a small 3 × 3 × 3 image array and extract its three channels.


```python
image_arr = np.array([
        [[10, 20, 30], [40, 50, 60], [70, 80, 90]],
        [[15, 25, 35], [45, 55, 65], [75, 85, 95]],
        [[20, 30, 40], [50, 60, 70], [80, 90, 100]]
        ])

print("Red:", image_arr[:,:,0])
print("Green:", image_arr[:,:,1])
print("Blue:", image_arr[:,:,2])
```
