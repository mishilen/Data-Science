## Random Module
<p>The <code>numpy.random</code> module supplements the built-in Python random module with functions for effciently generating whole arrays of sample values from many kinds of probability distributions. For example, you can get a 4 × 4 array of samples from the standard normal distribution using
<code>numpy.random.standard_normal</code></p>

```python
np.random.standard_normal(size=(4, 4))
```

<p>using the <code>random.default_rng<code></p>

```python
rng = np.random.default_rng()

np.random.default_rng().integers(5, 10), np.random.default_rng(seed=12345).integers(5,10)
```

<p>These random numbers are not truly random (rather, pseudorandom) but instead are generated by a confgurable random number generator that determines deterministically what values are created. Functions like numpy.random.standard_normal use the numpy.random module's default random number generator, but your code can be confgured to use an explicit generator</p>

```python
rng = np.random.default_rng(seed=12345)
ran0 = rng.standard_normal(size=(2,3))
```

```python
ran0
```

<p>Using <code>random.integers(low, high, size)</code> </p>

```python
rng.integers(5, 25, size=(4, 4))
```
