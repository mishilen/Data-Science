# 1. Vectorization
<p>Replacing explicit loops with array expressions is referred to by some people as vectorization.</p>

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.arange(-5, 5, 0.1)
data.size
```

<p><code>numpy.meshgrid</code> function takes two <b>one-dimensional</b> arrays and produces two <b>two-dimensional</b> matrices corresponding to all pairs of (x, y) in the two arrays</p>

```python
xs, ys = np.meshgrid(data, data)
xs, ys
```

## Making a title, plotting, and displaying

```python
plt.title("My first Plot")
plt.plot(xs, ys, "go-", label="y")
plt.show()
```

# 2. Expressing Conditional Logic as Array Operations
<p><code>numpy.where</code>function is a vectorized version of the ternary expression <code>x if condition else y</code> .</p>

```python
xarr = np.array([1.1, 1.2, 1.3, 1.4, 1.5])
yarr = np.array([2.1, 2.2, 2.3, 2.4, 2.5])
cond = np.array([True, False, True, True, False])
```

Suppose we wanted to take a value from xarr whenever the corresponding value in cond is True, and otherwise take the value from yarr .
A list comprehension doing this might look like:

```python
result = [(x if c else y) for x, y, c in zip(xarr, yarr, cond)]
result
```

<p>Using Numpy <code>where</code> function

```python
result = np.where(cond, xarr, yarr)
result
```

<p>The second and third arguments to numpy.where don’t need to be arrays; one or both of them can be
scalars.</p>

```python
my_arr = np.array([1, 5, 2, 9, 10, -1, 0, -2, -5])
my_arr
```

```python
# turning values from negative to positive

np.where(my_arr > 0, my_arr, my_arr * -1)
```

# 5. File Input and Output with Arrays

<p>NumPy is able to save and load data to and from disk in some text or binary formats.</p>
<p>numpy.save and numpy.load are the two workhorse functions for effciently saving and loading array
data on disk.</p>
<br>
<p> Arrays are saved by default in an uncompressed raw binary format with fle extension .npy</p>

```python
arr = np.arange(1, 101, 2).reshape(5, 10)
arr
```

```python
np.save("my_file", arr)
```

<p>If the fle path does not already end in .npy, the extension will be appended. The array on disk can then
be loaded with numpy.load</p>

```python
np.load("./my_file.npy")
```
